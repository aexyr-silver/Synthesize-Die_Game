```python
def writeDataIntoFile(data, filename, filetype='.txt', destination="data"):
  if filetype[0] != '.':
    filetype = '.' + filetype

  if filename[-len(filetype):] == filetype[-len(filetype):]:
    ### Already has filetype written at the end of the file name ###
    filepath = filename
    filetype = filename[-len(filetype):]

    if filetype[0] != '.':
      filetype = '.' + filetype

  else:
    ### Have NOT added filetype to the end of the file name ###
    filepath = filename + filetype

  directory = os.path.join(destination, filepath)

  if not os.path.exists(os.path.dirname(directory)):
    os.makedirs(os.path.dirname(directory))

  with open(directory, 'w') as file:
    if filetype == '.json':
      J.dump(data, file, indent=2)
    else:
      file.write(data)
```

### Establishing the Card Registry data based on Rarity weight.
Base      (x50),
Common    (x20),
Rare      (x10),
Historic  (x8),
Fabled    (x5),
Mythic    (x2),
Lost      (x1)

```python
def loadRegistry(data) -> list:
  registry_d = []

  for card_key, card_content in data.items():
    if card_content['rarity'] == 'Base':
      registry_d = addDuplicates(card_key, registry_d, 50)

    elif card_content['rarity'] == 'Common':
      registry_d = addDuplicates(card_key, registry_d, 20)

    elif card_content['rarity'] == 'Rare':
      registry_d = addDuplicates(card_key, registry_d, 10)

    elif card_content['rarity'] == 'Historic':
      registry_d = addDuplicates(card_key, registry_d, 8)

    elif card_content['rarity'] == 'Fabled':
      registry_d = addDuplicates(card_key, registry_d, 5)

    elif card_content['rarity'] == 'Mythic':
      registry_d = addDuplicates(card_key, registry_d, 2)

    elif card_content['rarity'] == 'Lost':
      registry_d = addDuplicates(card_key, registry_d, 1)

    else:
      print(f"ERROR: Card {card_content['name']} has an unspecified card rarity. Skipping card...")

  return registry_d
```

### Pre-Defined Libraries

```python
def totalMatches(dice=TRAD_DICE):
  print(f"Rolling left: {dice}")
  if (len(dice) == 1): return 1

  curr_side = dice[-1]
  print(f"  Currently checking tier {len(dice)}, d{curr_side}")

  curr_dice = dice[:-1]
  levels = 1 + curr_side * totalMatches(curr_dice)

  print(f"  Tier {len(dice)} returned {levels}.")
  return levels

# print(totalMatches())
```

### Game Engine Module
This is how the game will parse any match's results.

```python
def resolveMatch(player_card, opp_card) -> tuple:
  player = player_card['effect']
  opponent = opp_card['effect']

  total_points = 0
  opposing_points = 0

  if player['enable_award'] and opponent['opp_enable_award']:
    total_points += player['award'] + opponent['opp_award']

  if player['enable_remove'] and opponent['opp_enable_remove']:
    total_points -= player['remove'] + opponent['opp_remove']

  if opponent['enable_award'] and player['opp_enable_award']:
    opposing_points += opponent['award'] + player['opp_award']

  if opponent['enable_remove'] and player['opp_enable_remove']:
    opposing_points -= opponent['remove'] + player['opp_remove']

  total_points = total_points * player['multiply'] * opponent['opp_multiply']
  opposing_points = opposing_points * opponent['multiply'] * player['opp_multiply']

  return (player_card, opp_card), (total_points, opposing_points)
```

### Saving the Match Results.

```python
def saveMatchResults(cardUsed, matchResult, player_profile):
  print(f"Saving match results... Gained {matchResult[0]} points.")
  player_profile['points'] += matchResult[0]
  print(f"  Current total points: {player_profile['points']} points.\n")

  player_dict = cardUsed[0]
  opponent_dict = cardUsed[1]

  print(f"Winner of match being calculated...")
  if matchResult[0] > matchResult[1]:
    # Player won against Opponent
    print(f"  Congratulations! {player_profile['name']} won.")
    win_dict = player_dict
  else:
    # Opponent won against Player, or it's a tie.
    print(f"  ... It doesn't matter! {player_profile['name']} lost.")
    win_dict = opponent_dict

  player_key = None
  opponent_key = None
  win_key = None

  for key, card_dict in card_data.items():
      if card_dict is win_dict:
          win_key = key

      elif card_dict is player_dict:
          player_key = key

      elif card_dict is opponent_dict:
          opponent_key = key

  if win_key not in player_profile['cardsUnlocked']:
    print(f"  You have scanned the winning side card.\n  {win_dict['name']} has been added to your collection.\n")
    player_profile['cardsUnlocked'].append(win_key)

  ### Add match result into the player_profile ###
  print(f"\nInitializing synthesis process...")
  synthesis = player_profile['diceSynthesis']
  curr_tier = player_profile['currentTier']

  print(f"Current Tier: {curr_tier}...")
  if curr_tier == 1:
    TIERS.get(curr_tier)['name'].append(win_key)

  else:
    TIERS.get(curr_tier)['name'].append({
        "tier": curr_tier,
        "id": len(TIERS.get(curr_tier)),
        "win_card": win_key
    })

  print(f"  Successfully added \'{card_data[win_key]['name']}\' as the winning side card of this match.")

  player_profile = synthesize(player_profile, curr_tier)

  return curr_tier, player_profile
```

```python
def synthesize(player_profile, curr_tier):
  synthesis = player_profile['diceSynthesis']

  if curr_tier >= 7:
    return player_profile

  ### Appending the right tier is done before this function should be called ###
  # This function will only resolve the bottom-up synthesis operations to save #
  # and begin the next tier. #

  global TRAD_DICE

  if len(TIERS.get(curr_tier)) >= TRAD_DICE[curr_tier]:
    if curr_tier == 1:
      player_profile['currentCards'].append(TIERS.get(curr_tier).pop())
    else:
      TIERS.get(curr_tier + 1).append(
          {
              "tier" : curr_tier + 1,
              "id" : len(TIERS.get(curr_tier + 1)),
              "win_card" : None
           }
      )

    TIERS.get(curr_tier).clear()
    player_profile['currentTier'] = curr_tier + 1
    player_profile['currentCards'] = []

    for curr_tier in range(1, player_profile['currentTier'] + 1):
      player_profile['currentCards'].append(TIERS.get(curr_tier).pop())

    return synthesize(player_profile, curr_tier + 1)

  return synthesize(player_profile, curr_tier + 1)
```
